```python
import pandas as pd

df_hits = pd.read_parquet('data/diploma_main_dataset/ga_hits.parquet.gz')
df_sessions = pd.read_parquet('data/diploma_main_dataset/ga_session.parquet.gz')
```

## Проверка пропусков

```python
df_sessions.isna().sum()
```

```python
df_hits.isna().sum()
```

```python
for col in df_sessions.columns:
    if df_sessions[col].isna().sum() > 0:
        if df_sessions[col].dtype == 'object':
            most_freq = df_sessions[col].mode()[0]
            df_sessions[col].fillna(most_freq, inplace=True)
        else:
            mean_value = df_sessions[col].mean()
            df_sessions[col].fillna(mean_value, inplace=True)
```

```python
for col in df_hits.columns:
    if df_hits[col].isna().sum() > 0:
        if df_hits[col].dtype == 'object':
            most_freq = df_hits[col].mode()[0]
            df_hits[col].fillna(most_freq, inplace=True)
        else:
            mean_value = df_hits[col].mean()
            df_hits[col].fillna(mean_value, inplace=True)
```

## Преобразование дат и времени

```python
df_sessions['visit_date'] = pd.to_datetime(df_sessions['visit_date'], errors='coerce')
df_sessions['visit_time'] = pd.to_datetime(df_sessions['visit_time'], format='%H:%M:%S').dt.time

df_hits['hit_date'] = pd.to_datetime(df_hits['hit_date'], errors='coerce')
```
