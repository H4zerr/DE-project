```python
import pandas as pd
from matplotlib import pyplot as plt

df_hits = pd.read_parquet('data/diploma_main_dataset/ga_hits.parquet.gz')
df_sessions = pd.read_parquet('data/diploma_main_dataset/ga_session.parquet.gz')
```

## Sessions

```python
plt.figure(figsize = (10,5))
df_sessions['visit_date'] = pd.to_datetime(df_sessions['visit_date'])
df_sessions['visit_date'].value_counts().sort_index().plot(kind='line')
plt.title('Динамика количества визитов по датам')
plt.xlabel('Дата визита')
plt.ylabel('Количество визитов')
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize = (10,5))
df_sessions['device_category'].value_counts().sort_index().plot(kind='bar')
plt.title('Распределение по типу устройств')
plt.xlabel('Тип устройства')
plt.ylabel('Количество визитов')
plt.show()
```

```python
plt.figure(figsize = (10,5))
df_sessions['geo_city'].value_counts().head(10).plot(kind='bar')
plt.title('Топ 10 городов по количеству визитов')
plt.xlabel('Город')
plt.ylabel('Количество визитов')
plt.show()
```

```python
plt.figure(figsize = (8,4))
df_sessions['utm_source'].value_counts().head(10).plot(kind='bar')
plt.title('Топ 10 источников трафика')
plt.xlabel('Источник')
plt.ylabel('Количество визитов')
plt.show()
```

## Hits

```python
plt.figure(figsize=(10,5))
df_hits['event_category'].value_counts().head(10).plot(kind='bar')
plt.title('ТОП 10 категорий событий')
plt.xlabel('Категория события')
plt.ylabel('Количество')
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(10,5))
df_hits['hit_date'] = pd.to_datetime(df_hits['hit_date'])
df_hits['hit_date'].value_counts().sort_index().plot(kind='line')
plt.title('Активность событий по датам')
plt.xlabel('Дата')
plt.ylabel('Количество событий')
plt.grid(True)
plt.show()
```
